```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# Load data
with open('../samples_alt_trv/L1/L1_L00_outs/L1_L00_AT4G14210_1_4_exons_lengths.txt') as f:
    len_1_4 = [int(line.strip()) for line in f if line.strip()]
    
with open('../samples_alt_trv/L1/L1_L00_outs/L1_L00_AT4G14210_5_15_exons_lengths.txt') as f:
    len_5_15 = [int(line.strip()) for line in f if line.strip()]

with open('../samples_alt_trv/L1/L1_L00_outs/L1_L00_TRV-PDS3_lengths.txt') as f:
    len_trv_pds3 = [int(line.strip()) for line in f if line.strip()]
    
with open('../samples_alt_trv/L1/L1_L00_outs/L1_L00_TRV-dGAG_lengths.txt') as f:
    len_trv_dgag = [int(line.strip()) for line in f if line.strip()]

with open('../samples_alt_trv/L1/L1_L00_outs/L1_L00_AT2G37620_exons_lengths.txt') as f:
    len_norm = [int(line.strip()) for line in f if line.strip()]

# Load data
with open('../samples_alt_trv/L2/L2_L00_outs/L2_L00_AT4G14210_1_4_exons_lengths.txt') as f:
    len_1_4_l2 = [int(line.strip()) for line in f if line.strip()]
    
with open('../samples_alt_trv/L2/L2_L00_outs/L2_L00_AT4G14210_5_15_exons_lengths.txt') as f:
    len_5_15_l2 = [int(line.strip()) for line in f if line.strip()]

with open('../samples_alt_trv/L2/L2_L00_outs/L2_L00_TRV-PDS3_lengths.txt') as f:
    len_trv_pds3_l2 = [int(line.strip()) for line in f if line.strip()]
    
with open('../samples_alt_trv/L2/L2_L00_outs/L2_L00_TRV-dGAG_lengths.txt') as f:
    len_trv_dgag_l2 = [int(line.strip()) for line in f if line.strip()]

with open('../samples_alt_trv/L2/L2_L00_outs/L2_L00_AT2G37620_exons_lengths.txt') as f:
    len_norm_l2 = [int(line.strip()) for line in f if line.strip()]
```

```python
# Set style
sns.set(style='whitegrid', palette='pastel')
plt.figure(figsize=(10, 12))

# Plot histograms
plt.subplot(5, 2, 1)
sns.histplot(len_1_4, bins=50, color='red', kde=True)
plt.title('L1 AT4G14210_1_4 (TRV-PDS3 target)')
plt.xlabel('Read Length (nt)')
plt.ylabel('Count')
plt.axvline(np.mean(len_1_4), color='darkred', linestyle='--', label=f'Mean: {np.mean(len_1_4):.1f}')
plt.legend()

# Plot histograms
plt.subplot(5, 2, 2)
sns.histplot(len_1_4_l2, bins=50, color='blue', kde=True)
plt.title('L2 AT4G14210_1_4 (TRV-PDS3 target)')
plt.xlabel('Read Length (nt)')
plt.ylabel('Count')
plt.axvline(np.mean(len_1_4_l2), color='darkblue', linestyle='--', label=f'Mean: {np.mean(len_1_4_l2):.1f}')
plt.legend()

plt.subplot(5, 2, 3)
sns.histplot(len_5_15, bins=50, color='red', kde=True)
plt.title('L1 AT4G14210_5_15 (Native)')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_5_15), color='darkred', linestyle='--', label=f'Mean: {np.mean(len_5_15):.1f}')
plt.legend()

plt.subplot(5, 2, 4)
sns.histplot(len_5_15_l2, bins=50, color='blue', kde=True)
plt.title('L2 AT4G14210_5_15 (Native)')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_5_15_l2), color='darkblue', linestyle='--', label=f'Mean: {np.mean(len_5_15_l2):.1f}')
plt.legend()

plt.subplot(5, 2, 5)
sns.histplot(len_trv_pds3, bins=50, color='red', kde=True)
plt.title('L1 TRV-PDS3 transcript')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_trv_pds3), color='darkred', linestyle='--', label=f'Mean: {np.mean(len_trv_pds3):.1f}')
plt.legend()

plt.subplot(5, 2, 6)
sns.histplot(len_trv_pds3_l2, bins=50, color='blue', kde=True)
plt.title('L2 TRV-PDS3 transcript')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_trv_pds3_l2), color='darkblue', linestyle='--', label=f'Mean: {np.mean(len_trv_pds3_l2):.1f}')
plt.legend()

plt.subplot(5, 2, 7)
sns.histplot(len_trv_dgag, bins=50, color='red', kde=True)
plt.title('L1 TRV-dGAG transcript')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_trv_dgag), color='darkred', linestyle='--', label=f'Mean: {np.mean(len_trv_dgag):.1f}')
plt.legend()

plt.subplot(5, 2, 8)
sns.histplot(len_trv_dgag_l2, bins=50, color='blue', kde=True)
plt.title('L2 TRV-dGAG transcript')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_trv_dgag_l2), color='darkblue', linestyle='--', label=f'Mean: {np.mean(len_trv_dgag_l2):.1f}')
plt.legend()

plt.subplot(5, 2, 9)
sns.histplot(len_norm, bins=50, color='red', kde=True)
plt.title('L1 ACT1 transcript')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_norm), color='darkred', linestyle='--', label=f'Mean: {np.mean(len_norm):.1f}')
plt.legend()

plt.subplot(5, 2, 10)
sns.histplot(len_norm_l2, bins=50, color='blue', kde=True)
plt.title('L2 ACT1 transcript')
plt.xlabel('Read Length (nt)')
plt.axvline(np.mean(len_norm_l2), color='darkblue', linestyle='--', label=f'Mean: {np.mean(len_norm_l2):.1f}')
plt.legend()

plt.tight_layout()
# plt.savefig('read_length_distribution.png', dpi=300)
plt.show()

# Print statistics
# print(f"1_4 Stats: Mean={np.mean(len_1_4):.1f}, Median={np.median(len_1_4)}, Min={min(len_1_4)}, Max={max(len_1_4)}")
# print(f"5_15 Stats: Mean={np.mean(len_5_15):.1f}, Median={np.median(len_5_15)}, Min={min(len_5_15)}, Max={max(len_5_15)}")
```
